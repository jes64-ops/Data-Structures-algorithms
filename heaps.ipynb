```python
#Build Min Heap (heapify)

#time O(n) space : O(1)

A = [ -4 , 3, 1, 0 , 2, 5, 10 ,8 , 12 , 9]

import heapq
heapq.heapify(A)

A
```

```python
# heap Push (insert elements)
 # time : O(log n )
heapq.heappush(A, 4)
A
```

```python
# heap Pop (Extract min)
#time O(log n )

minn = heapq.heappop(A)

A, minn
```

```python
#heap sort
#time O(n log n ) ,space O (n)

def heapsort(arr):
  heapq.heapify(arr)
  n = len(arr)
  new_list = [0]* n

  for i in range(n):
    minn = heapq.heappop(arr)
    new_list[i] = minn

  return new_list

heapsort([1,3,5,7,9,2,4,6,8,0])
```

```python
# max heap

B = [-1, 2,3,4 ,5,7,6,8,9,]
n= len(B)

for i in range(n):
  B[i] = -B[i]
heapq.heapify(B)

B
```

```python
largest = -heapq.heappop(B)

largest
```

```python
heapq.heappush(B , -7)

B
```
