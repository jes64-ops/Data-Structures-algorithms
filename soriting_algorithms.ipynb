```python
#Bubble sort
#time o(n^2)
#Space O(1)

A = [-5 , 3, 2 , 1 , -3, -3 , 7 ,2 , 2]

def bubble_sort(arr):
  n = len(arr)
  flag = True
  while flag:
    flag = False
    for i in range(1,n):
      if arr[i-1]> arr[i]:
        flag = True
        arr[i-1], arr[i] = arr[i], arr[i-1]

bubble_sort(A)
A
```

```python
#Insertion sorting
#time O(n^2 )
#space O(1)

B = [-5 , 3, 2 , 1 , -3, -3 , 7 ,2 , 2]

def insertion_sort(arr):
  n= len(arr)
  for i in range(1,n):
    for j in range(i , 0 , -1):
      if arr[j-1]> arr[j]:
        arr[j-1], arr[j] = arr[j], arr[j-1]
      else:
        break
insertion_sort(B)
B
```

```python
#Selection sort
# time: O(n^2)
#space O(1)
C = [-5 , 3, 2 , 1 , -3, -3 , 7 ,2 , 2]

def selection_sort(arr):
  n = len(arr)
  for i in range(n):
    min_index = i
    for j in range(i+1 , n):
      if arr[j]< arr[min_index]:
        min_index = j
      arr[i], arr[min_index]= arr[min_index], arr[i]

selection_sort(C)
C
```

```python
# Merge Sort
D = [-5, 3, 2, 1, -3, -3, 7, 2, 2]

def merge_sort(arr):
    n = len(arr)

    # Base case
    if n <= 1:
        return arr

    # Split the array
    m = n // 2
    L = arr[:m]
    R = arr[m:]

    # Recursively sort each half
    L = merge_sort(L)
    R = merge_sort(R)

    # Merge the two sorted halves
    sorted_arr = [0] * n

    l = 0          # Pointer for left array
    r = 0          # Pointer for right array
    i = 0          # Pointer for merged array

    L_len = len(L)
    R_len = len(R)

    while l < L_len and r < R_len:
        if L[l] <= R[r]:
            sorted_arr[i] = L[l]
            l += 1
        else:
            sorted_arr[i] = R[r]
            r += 1
        i += 1

    # Copy remaining elements from L
    while l < L_len:
        sorted_arr[i] = L[l]
        l += 1
        i += 1

    # Copy remaining elements from R
    while r < R_len:
        sorted_arr[i] = R[r]
        r += 1
        i += 1

    return sorted_arr


print(merge_sort(D))
```

```python
# quick sort
# time O(nlog n )
#space O(n)'
E = [-5, 3, 2, 1, -3, -3, 7, 2, 2]

def quick_sort(arr):
  if len(arr) < 1:
    return arr

  # p = piovot

  p = arr[-1]

  L = [x for x in arr[:-1] if x <= p]
  R = [x for x in arr[:-1] if x > p]

  L = quick_sort(L)
  R = quick_sort(R)

  return L + [p] + R
quick_sort(E)
```

```python
# Counting sort
# time O(n +k )
# k is the max of the array


F = [5, 3, 2, 1, 3, 3, 7, 2, 2]

def counting_sort(arr):
  n = len(arr)
  maxx = max(arr)
  counts = [0] * (maxx + 1)

  for x in arr:
    counts[x] += 1
  i = 0
  for c in range(maxx + 1):
    while counts[c] > 0:
      arr[i] = c
      i += 1
      counts[c] -= 1

counting_sort(F)
F
```
